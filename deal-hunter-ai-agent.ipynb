**Project Description:** Multi-Agent Pricing and Deal Recommendation System (Google ADK v1.19.0)

This project implements a multi-agent intelligent shopping and deal-recommendation pipeline using the Google Agent Development Kit (ADK) **v1.19.0.**
The system is designed to analyze a user’s product query and automatically generate the best purchasing recommendation by combining information retrieved from multiple specialized agents.

The pipeline integrates parallel tool-assisted agents, structured knowledge synthesis, and sequential decision orchestration, and is fully functional inside an end-to-end Kaggle Notebook environment.

**1. Motivation and Objective**

Modern e-commerce consumers often face challenges in identifying the most cost-effective purchasing options due to the large number of retailers, marketplaces, and available discount sources.
This project aims to solve this challenge by automating the process of:

Retrieving new product prices

Retrieving used/refurbished marketplace offers

Fetching active store discount codes

Analyzing all available options

Providing a clear, data-grounded recommendation

The goal is to create a scalable multi-agent framework that elevates decision-making for shoppers using the power of LLM-based intelligence.

**2. System Architecture Overview**

The solution is structured into five conceptual layers, forming a robust multi-agent research and recommendation engine.

**2.1 Tool Layer (Data Retrieval Functions)**
Three Python functions simulate external data sources:

**Retail price search** (Amazon, Best Buy, Walmart)

**Used/refurb marketplace search** (eBay, Poshmark, Mercari)

**Coupon retrieval** (e.g., “SAVE20” for Walmart)

Each tool is wrapped using AgentTool to make it available to LLM agents via Google ADK.

**2.2 Specialized LLM Agents (Parallel Processing)**

Three domain-oriented agents operate independently:

**Retail Specialist** → Fetches new product prices

**Marketplace Specialist** → Fetches used/refurb options

**Coupon Specialist** → Retrieves relevant discounts

These agents use their respective tools to retrieve structured price information based on the user’s query.

To optimize performance, the system uses ParallelAgent to execute all three sub-agents simultaneously

**2.3 Aggregation and Reasoning Layer (DealHunter Agent)**

After the parallel agents complete execution, the DealHunter agent synthesizes their outputs using a structured prompt.

**This agent:**


* Reads all collected pricing and discount data
* Compares new vs. used options
* Applies coupon adjustments where appropriate
* Evaluates cost, value, and riskGenerates a final purchasing recommendation


The output includes both a narrative explanation and a concise bullet-point summary.

**2.4 Orchestration Layer (SequentialAgent Pipeline)**

The entire workflow is managed through a SequentialAgent, which ensures:


* Parallel agents execute first and populate shared state variables.
* The DealHunter agent receives all required inputs.
* A unified recommendation is produced.


This design aligns with multi-step decision pipelines commonly used in industry-grade agentic systems.

**2.5 Execution Layer (InMemoryRunner)**

Google ADK’s InMemoryRunner is used to execute the multi-agent pipeline.
This runner handles:


* Session tracking*
* Conversation state
* Agent invocation
* Response aggregation

The user interacts with the system by providing a natural-language query such as:

“**Find me the best deal for Sony WH-1000XM5.”**

The runner returns a fully synthesized answer generated by the LLM.

```python
code = r'''
from google.adk.agents import LlmAgent, ParallelAgent, SequentialAgent
from google.adk.tools import AgentTool

# ==========================================
# 1. TOOLS (Python functions)
# ==========================================

def search_retailers(product_name: str) -> str:
    """
    Mock search for new prices at major retailers.
    """
    print(f"[RetailSpecialist] Searching retailers for: {product_name}")
    return """\
- Amazon: Sony WH-1000XM5 - $348.00 (In Stock)
- Best Buy: Sony WH-1000XM5 - $349.99 (In Stock)
- Walmart: Sony WH-1000XM5 - $329.00 (Sale)
    """
search_retailers.name = "search_retailers"
search_retailers.description = search_retailers.__doc__.strip()


def search_marketplaces(product_name: str) -> str:
    """
    Mock search for used / refurbished prices.
    """
    print(f"[MarketplaceSpecialist] Searching marketplaces for: {product_name}")
    return """\
- eBay: Sony WH-1000XM5 (Refurbished) - $199.99 (Trusted Seller)
- Poshmark: Sony WH-1000XM5 (Used - Like New) - $180.00
- Mercari: Sony WH-1000XM5 (Open Box) - $210.00
    """
search_marketplaces.name = "search_marketplaces"
search_marketplaces.description = search_marketplaces.__doc__.strip()


def find_coupons(store_names: str) -> str:
    """
    Mock search for coupons for specific stores.
    """
    print(f"[CouponSpecialist] Looking for coupons for: {store_names}")
    return """\
- Walmart: Code 'SAVE20' (20% off select electronics) - Verified today.
- eBay: No active codes for electronics.
    """
find_coupons.name = "find_coupons"
find_coupons.description = find_coupons.__doc__.strip()


# Wrap functions as ADK tools
search_retailers_tool = AgentTool(search_retailers)
search_marketplaces_tool = AgentTool(search_marketplaces)
find_coupons_tool = AgentTool(find_coupons)

# ==========================================
# 2. AGENTS
# ==========================================

MODEL_NAME = "gemini-2.0-flash"

# Retail specialist
retail_agent = LlmAgent(
    name="RetailSpecialist",
    model=MODEL_NAME,
    instruction=(
        "You are a retail search expert. "
        "Use the `search_retailers` tool to find current prices for NEW items. "
        "The user will mention a product name. "
        "Call the tool with that product name. "
        "Return a concise bullet list of options."
    ),
    description="Finds prices for new products from major retailers.",
    tools=[search_retailers_tool],
    output_key="retail_offers",
)

# Marketplace specialist
marketplace_agent = LlmAgent(
    name="MarketplaceSpecialist",
    model=MODEL_NAME,
    instruction=(
        "You are a second-hand and refurbished market expert. "
        "Use `search_marketplaces` to find used or refurbished deals. "
        "Return a concise bullet list."
    ),
    description="Finds deals on used/refurb items.",
    tools=[search_marketplaces_tool],
    output_key="marketplace_offers",
)

# Coupon specialist
coupon_agent = LlmAgent(
    name="CouponSpecialist",
    model=MODEL_NAME,
    instruction=(
        "You are a coupon expert. "
        "Use `find_coupons` to get promo codes for relevant stores. "
        "Return a short bullet list."
    ),
    description="Finds active promo codes.",
    tools=[find_coupons_tool],
    output_key="coupon_offers",
)

# Parallel group
search_group = ParallelAgent(
    name="SearchGroup",
    sub_agents=[retail_agent, marketplace_agent, coupon_agent],
)

# Deal Hunter aggregator
deal_hunter_agent = LlmAgent(
    name="DealHunter",
    model=MODEL_NAME,
    instruction="""
You are The Deal Hunter.\n\nInputs:\n\n[RETAIL OFFERS]\n{retail_offers}\n\n[MARKETPLACE OFFERS]\n{marketplace_offers}\n\n[COUPON OFFERS]\n{coupon_offers}\n\nTasks:\n1. Identify best new price.\n2. Identify best used price.\n3. Apply coupons where relevant.\n4. Recommend ONE best deal overall.\n5. Provide 2–3 paragraph explanation and a bullet summary.\n""",
)

# Final pipeline
root_agent = SequentialAgent(
    name="root_agent",
    sub_agents=[search_group, deal_hunter_agent],
)
'''

with open("Capstone_project/agent.py", "w") as f:
    f.write(code)
```
